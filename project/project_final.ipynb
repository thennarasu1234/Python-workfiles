```python
import telebot
import requests

# === Your Keys ===
TELEGRAM_TOKEN = "8131160858:AAEhsL21lyHVs_qvrSn_4SGF3kTHr4xO79Q"
HF_TOKEN = "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"

# === API Setup ===
API_URL = "https://router.huggingface.co/v1/chat/completions"
headers = {"Authorization": f"Bearer {HF_TOKEN}"}

# === Initialize Bot ===
bot = telebot.TeleBot(TELEGRAM_TOKEN)

# Function to call Hugging Face API
def ask_huggingface(prompt):
    payload = {
        "messages": [{"role": "user", "content": prompt}],
        "model": "openai/gpt-oss-120b:scaleway"
    }
    response = requests.post(API_URL, headers=headers, json=payload)
    data = response.json()
    try:
        return data["choices"][0]["message"]["content"]
    except Exception as e:
        return "❌ Error: Could not fetch response."

# === Telegram Handlers ===

@bot.message_handler(commands=['start', 'help'])
def send_welcome(message):
    bot.reply_to(message, "👋 Hi! Send me any question, and I'll answer using Hugging Face models!")

@bot.message_handler(func=lambda message: True)
def chat_with_huggingface(message):
    user_input = message.text
    bot.send_chat_action(message.chat.id, 'typing')  # show typing...
    answer = ask_huggingface(user_input)
    bot.send_message(message.chat.id, answer)

# === Run the bot ===
print("🤖 Bot is running...")
bot.polling()
```
